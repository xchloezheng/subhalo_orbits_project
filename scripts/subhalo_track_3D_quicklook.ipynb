```python
# Loop from 0 up to (but not including) 281

x = []
y = []
z = []

for i in range(281):
    # print(i)
    filename = f"{i}-halos"

    with open(file_path+filename, 'r') as f:
        # print(f"Processing: {filename}")

        lines = f.readlines()

        if len(lines) == 2:
            # print(f"File {filename}")
        
            second_line = lines[1]
            second_line = second_line.strip()
            second_line_elements = second_line.split()

            # print(type(second_line_elements[5]))

            x.append(float(second_line_elements[5])-1e4)
            y.append(float(second_line_elements[6])-1e4)
            z.append(float(second_line_elements[7])-1e4)

# print(x)
# print(y)
# print(z)

# print(len(x))
# print(len(y))
# print(len(z))

# print(type(x[0]))
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection='3d')

# Plot the data as a 3D line
ax.plot3D(x, y, z, 'blue', marker='*', markersize=4)

# Set labels for the axes
ax.set_xlabel('X Value')
ax.set_ylabel('Y Value')
ax.set_zlabel('Z Value')
ax.set_title('3D Line Plot Over Time')

plt.show()
```
