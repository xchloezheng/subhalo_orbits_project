```python
import matplotlib.pyplot as plt
import numpy as np

files = ['./Data/host_halo/host_sigv_CDM.txt', './Data/host_halo/host_sigv_SIDM.txt', './Data/host_halo/host_sigv_subhalo_evp_3_1_1.txt', './Data/host_halo/host_sigv_static_NFW.txt', './Data/host_halo/host_sigv_hosthalo_3_1.txt']
labels = ['CDM', 'SIDM', 'Subhalo 3_1_1', 'Static NFW', 'Host Halo 3_1']
cols = [(1, r'$\rho(r)$'), (2, '$M(<r)$'), (3, '$\sigma_v(r)$')]

fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharex=True)

for f, lbl in zip(files, labels):
    data = np.loadtxt(f)
    radius = data[:, 0]
    for i, (col_idx, name) in enumerate(cols):
        axes[i].scatter(radius, data[:, col_idx], label=lbl,alpha=0.7)
        axes[i].set_title(name)
        axes[i].set_xlabel('Radius ($r$)')

axes[0].set_xscale('log')
axes[0].set_yscale('log') # Density often looks better on a log scale
axes[0].legend()
plt.tight_layout()
plt.show()
```
